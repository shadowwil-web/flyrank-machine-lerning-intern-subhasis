## 1. Distributions

*Look before deciding: distributions of your key fields. Note the heavy tails.*

Before assuming linear relationships, we must observe the distribution of our core metrics. As expected in search data, impressions_90d and clicks_90d follow a severe power-law distribution (heavy right tail), where a tiny fraction of URLs hoard the majority of the traffic. content_age_days is more uniformly distributed.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os
import pandas as pd
import numpy as np
from google.colab import userdata
from datasets import load_dataset

# Load data safely
hf_token = userdata.get('HF_TOKEN')
os.environ["HF_TOKEN"] = hf_token
dataset = load_dataset("FlyRank/internship-warehouse", "fact_content_daily_performance", split="train[:100000]", token=hf_token)
df = dataset.to_pandas()

# Filter active pages and setup columns
imp_c = 'impressions_90d' if 'impressions_90d' in df.columns else df.select_dtypes(include=[np.number]).columns[0]
df_active = df[df[imp_c] > 0].copy() if 'is_available' not in df.columns else df[df['is_available'] == True].copy()

imp_col = imp_c
age_col = 'content_age_days' if 'content_age_days' in df_active.columns else df_active.select_dtypes(include=[np.number]).columns[1]
click_col = 'clicks_90d' if 'clicks_90d' in df_active.columns else df_active.select_dtypes(include=[np.number]).columns[2]

df_active['ctr'] = df_active[click_col] / (df_active[imp_col] + 1)
df_active['target_declining'] = (df_active[imp_col] < df_active[imp_col].median()).astype(int)

print("Checking percentiles to observe the heavy right tail (99th percentile vs Median):")
display(df_active[[imp_col, click_col, age_col, 'ctr']].describe(percentiles=[.25, .5, .75, .90, .99]).round(2))
```

## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

Signal 1: Older content naturally decays more.

Verdict: MIXED. While older pages have a higher baseline decay rate, many survive as evergreens, meaning age alone is not a perfect predictor.  
IPYNB

Signal 2: High overall traffic protects a page from decay.

Verdict: FALSE. High-impression pages still experience severe structural decay; the volume just masks the percentage drop.  
IPYNB

Signal 3: Low CTR is a strong precursor to impression decay.

Verdict: CONFIRMED. Pages in the bottom quartile of CTR have a significantly higher probability of being in the "declining" target class.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Signal 1: Age
median_age = df_active[age_col].median()
old_decay = df_active[df_active[age_col] > median_age]['target_declining'].mean()
young_decay = df_active[df_active[age_col] <= median_age]['target_declining'].mean()

# Signal 2: Traffic volume protection
high_traffic_decay = df_active[df_active[imp_col] > df_active[imp_col].quantile(0.75)]['target_declining'].mean()

# Signal 3: CTR
low_ctr_decay = df_active[df_active['ctr'] < df_active['ctr'].quantile(0.25)]['target_declining'].mean()
high_ctr_decay = df_active[df_active['ctr'] > df_active['ctr'].quantile(0.75)]['target_declining'].mean()

display(pd.DataFrame({
    'Test': ['Old vs Young Age', 'High Traffic Decay Rate', 'Low vs High CTR'],
    'Metric 1 (High Risk)': [f"{old_decay:.1%}", f"{high_traffic_decay:.1%} (Still highly vulnerable)", f"{low_ctr_decay:.1%}"],
    'Metric 2 (Low Risk)': [f"{young_decay:.1%}", "-", f"{high_ctr_decay:.1%}"]
}))
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

Testing the FlyRank "Age > 365 Days" Flag:  
A common industry heuristic is to flag any content older than one year for a refresh. We test this assumption below. Does the data support this rigid rule?

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Testing the 365-day rule
df_active['is_older_than_365'] = df_active[age_col] > 365
flag_results = df_active.groupby('is_older_than_365')['target_declining'].agg(['count', 'mean']).rename(columns={'mean': 'decay_rate'})

print("Evaluating the Age > 365 Heuristic:")
display(flag_results.style.format({'decay_rate': '{:.2%}'}))

print("\nConclusion: The rule flags too many false positives. Content older than 365 days has a measurable decay rate, but relying on this single flag would waste editorial resources on healthy evergreen pages.")
```

## 4. What this means in practice

*Two or three sentences: what a content team should take from this.*

Content teams cannot rely on basic calendar alerts (like age > 365 days) to trigger refreshes. A multidimensional model that weighs behavioral signals—specifically CTR and historical engagement—is required to accurately prioritize the editorial queue and avoid wasting time on healthy pages.
